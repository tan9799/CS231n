In particular, an issue commonly seen with activation functions is getting zero (or close to zero) gradient flow during backpropagation. Which of the following activation functions have this problem? If you consider these functions in the one dimensional case, what types of input would lead to this behaviour?
1. Sigmoid
2. ReLU
3. Leaky ReLU

$\color{blue}{\textit Your Answer:}$ 
Sigmoid： 当输入是很大的正数或负数时，输出接近 1 或 0，进入饱和区。其导数为 σ(x)(1−σ(x))，此时接近零，容易导致梯度消失。

ReLU： 当输入小于 0 时，输出和导数都为零，会阻断梯度传播；当输入大于 0 时，导数为 1。输入等于 0 时不可导，实现中通常将导数设为 0。

Leaky ReLU： 当输入小于 0 时，导数为一个较小的正常数 α；输入大于 0 时，导数为 1。因此它避免了 ReLU 在负半轴上梯度完全为零的问题，但如果 α 很小，负半轴上的梯度仍可能接近零。
